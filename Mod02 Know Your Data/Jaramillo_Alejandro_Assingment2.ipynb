```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('./Income Data.CSV')
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
def fns(column):
    #find 5 num summary
    min = np.min(df[column])
    Q1 = np.percentile(df[column], 25)
    median = np.median(df[column])
    Q3 = np.percentile(df[column], 75)
    max = np.max(df[column])

    #find mean
    mean = df[column].mean()


    #Print stats
    print(f'------------ FNS for {column.capitalize()} -------------')
    print(f"Min: {min}")
    print(f"Q1: {Q1}")
    print(f"Median: {median}")
    print(f"Q3: {Q3}")
    print(f"Max: {max}")
    print(f"Mean: {mean}")

fns("income")
fns("education")
fns("prestige")
    
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
def histogram(column):
    print(f"Histogram of {column}")
    plt.hist(df[column])
    plt.show()  

histogram("prestige")
histogram("education")
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
def ScatterPlot (Col_x, Col_y):
    print(f"ScatterPlot for {Col_x} compared to {Col_y}")
    plt.scatter(x= df[Col_x], y = df[Col_y])
    plt.title(f"{Col_x} compared to {Col_y}")
    plt.xlabel(Col_x)
    plt.ylabel(Col_y)
    plt.show()

ScatterPlot("income", "education")
ScatterPlot("education", "prestige")
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python

def variance(column):
    std = df[column].std()
    print(f"Standard Deviation: {std:.2f}")
    variance = std ** 2 
    print(f"Variance:{variance:.2f}")

variance("income")

```
